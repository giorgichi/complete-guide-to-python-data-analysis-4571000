## 01_03_01 Dictionaries

```python
# key-value pairs
capitals = {'United States': 'Washington, DC', 'France': 'Paris', 'Italy': 'Rome'}
```

```python
capitals
```

```python
len(capitals), len({})
```

```python
capitals['Italy']
```

```python
# Adding a new key-value pair to the dictionary
capitals['Spain'] = 'Madrid'
```

```python
capitals
```

```python
capitals['Germany']
```

```python
'Germany' in capitals, 'Italy' in capitals
```

```python
# Combining two dictionaries using `**` unpacking operator
# if keys are repeated the last one is used (i.e. value of the last key overrides the previous one)
morecapitals = {'Germany': 'Berlin', 'United Kingdom': 'London', 'United States': 'DC'}
{**capitals, **morecapitals}
```

```python
capitals.update(morecapitals)
```

```python
capitals
```

```python
del capitals['United Kingdom']
```

```python
capitals
```

Keys of the dictionaries do not need to be strings (like names of countries above), we can also use number, or anything that is **hashable**.  
> **Hashable** means that Python can convert the value into a number (includes strings, numbers and tuples, but not lists)

```python
birthdays = {(7,15): 'Michele', (3,14): 'Albert'}
```

```python
birthdays[(7,15)]
```

```python
hash('Italy'), hash((7,15))
```

```python
hash([1, 2])
```

### Looping over the dictionaries

```python
for country in capitals:
    print(country)
```

```python
for country in capitals.keys():
    print(country)
```

```python
capitals.keys()
```

```python
list(capitals.keys())
```

```python
for capital in capitals.values():
    print(capital)
```

```python
for country, capital in capitals.items():
    print(country, capital)
```

```python
capitals.keys()
```

```python
list(capitals.keys())
```

```python
capitals.values()
```

```python
capitals.items()
```

```python
# Using defaultdict to set a default value for missing keys (i.e. keys that are not present in the dictionary)
capitals_default = collections.defaultdict(lambda: "I don't know!")
```

```python
# copy key-value pairs from capitals to capitals_default
capitals_default.update(capitals)
```

```python
capitals_default['Canada']
```

## 01_03_02 Sets

**Imputable** type of set of iterms that are **not duplicated** and **unordered** and **not indexed** (i.e. cannot get an individual element by index)

```python
# Even though we add Africa twice, it will only be stored once in the set because sets do not allow duplicate values.
continents = {'America', 'Europe', 'Asia', 'Oceania', 'Africa', 'Africa'}
```

```python
continents
```

```python
'Africa' in continents
```

```python
continents.add('Antarctica')
```

```python
continents.remove('Antarctica')
```

```python
for c in continents:
    print(c)
```

```python
continents = {'America', 'Europe', 'Asia', 'Oceania', 'Africa', 'Georgia'}
```

```python
continents
```

Sets can have **mixed elements**, but all elements should be hashable (includes, numbers, strings, tuples, but NOT lists)

```python
my_set = {1, 2, 3, 4, 'Gio', 'Nothing', (1, 2, 3)}
my_set
```

```python
for my in my_set:
    print(my)
```

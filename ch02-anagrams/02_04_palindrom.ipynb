```python
import collections
import dataclasses
```

```python
words = sorted({line.strip().lower() for line in open('words.txt', 'r')})
```

```python
def signature(word):
    return ''.join(sorted(word))

words_by_signature = collections.defaultdict(set)

for word in words:
    words_by_signature[signature(word)].add(word)
```

```python
# remove any signatures that only have one word associated with them, since those aren't anagrams
anagrams_by_signature = {sig: wordset for sig, wordset in words_by_signature.items() if len(wordset) > 1}
```

```python

import itertools

def palindromes(anagrams_by_signature):
    # Your code goes here.
    pals = []
    for words in [val for key, val in anagrams_by_signature.items() if len(key) > 6]:
        for word1, word2 in itertools.combinations(words, 2):
            if word1 == word2[::-1]:
                pals.append({word1, word2})


    return pals
```

```python
palindromes(anagrams_by_signature)
```
